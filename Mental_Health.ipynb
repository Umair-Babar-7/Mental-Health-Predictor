# 7. Feature Engineering

```python
top_countries = df['Country'].value_counts().index[:10].to_list()
```

```python
def group_countries(country):
  if country in top_countries:
    return country
  else:
    return 'Other'
```

```python
df['Grouped_Country'] = df['Country'].apply(group_countries)
```

```python
df['Grouped_Country'].value_counts()
```

# 8. Encoding Strategy

# 9. Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
skewed_col = ['Study_Hours']
other_numeric_cols = ['Age', 'Avg_Daily_Usage_Hours', 'Daily_Unlocks', 'Physical_Activity_Hours', 'Sleep_Hours_Per_Night']

ordinal_col = ['Stress_Level']
normal_cols = ['Gender', 'Academic_Level', 'Most_Used_Platform', 'Purpose_Of_Use', 'Grouped_Country']

feature_col = skewed_col + other_numeric_cols + ordinal_col + normal_cols
```

```python
X = df[feature_col]
y = df['Mental_Health_Score']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# 10. Preprocessing using ColumnTransformer

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
from sklearn import pipeline
# 1. Skewed Features
skewed_pipeline = Pipeline(steps=[
    ('log_transform', FunctionTransformer(np.log1p)),
    ('standard_scaler', StandardScaler())
])

# 2. Numeric Features
numeric_pipeline = Pipeline(steps=[
    ('standard_scaler', StandardScaler())
])

# 3. Ordinal Features
ordinal_pipeline = Pipeline(steps=[
    ('ordinal_encoding', OrdinalEncoder(categories=[['Low', 'Medium', 'High', 'Very High']]))
])

# 4. Nominal Features
nominal_pipeline = Pipeline(steps=[
    ('one_hot_encoding', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
preprocessor = ColumnTransformer(transformers=[
    ('skewed_transform', skewed_pipeline, skewed_col),
    ('numeric_transform', numeric_pipeline, other_numeric_cols),
    ('ordinal_transform', ordinal_pipeline, ordinal_col),
    ('nominal_transform', nominal_pipeline, normal_cols)
])
```

# 11. Build a Pipeline

# 12. Model Building

## 12.1 Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regression', LinearRegression())
])
lr_pipeline.fit(X_train, y_train)
lr_pred = lr_pipeline.predict(X_test)
lr_pred_train = lr_pipeline.predict(X_train)
```

## 12.2 Random Forest (default settings)

```python
from sklearn.ensemble import  RandomForestRegressor

rf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('random forest', RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(X_train, y_train)
```

```python
rf_pipeline_pred = rf_pipeline.predict(X_test)
rf_pipeline_train = rf_pipeline.predict(X_train)
```

## 12.3 Hyperparameter Tuning the Random Forest

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'random forest__n_estimators': [100, 200, 300],
    'random forest__max_depth': [5, 10, 15],
    'random forest__min_samples_split': [2, 5, 10],
    'random forest__min_samples_leaf': [1, 2, 4]
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

# 13. Model Evaluation

## 13.1 Linear Regression Evaluation

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

lr_r2 = r2_score(y_test, lr_pred)
lr_r2_train = r2_score(y_train, lr_pred_train)

lr_mae = mean_absolute_error(y_test, lr_pred)
lr_mse = mean_squared_error(y_test, lr_pred)

print(f'Accuracy of training dataset is {lr_r2_train}')
print(f'Accuracy of testing dataset is {lr_r2}')
print(f'Mean Absolute Error is {lr_mae}')
print(f'Mean Squared Error is {lr_mse}')
```

```python
# Create the plot
sns.regplot(x=y_test, y=lr_pred, scatter_kws={'alpha':0.5})

plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Model Accuracy: Actual vs. Predicted')
plt.show()
```

## 13.2 Random Forest Evaluation

```python
rf_r2_test = r2_score(y_test, rf_pipeline_pred)
rf_r2_train = r2_score(y_train, rf_pipeline_train)

rf_mae = mean_absolute_error(y_test, rf_pipeline_pred)
rf_mse = mean_squared_error(y_test, rf_pipeline_pred)

print(f'Accuracy of testing dataset is {rf_r2_test}')
print(f'Accuracy of training dataset is {rf_r2_train}')

print(f'Mean Absolute Error of testing dataset is {rf_mae}')
print(f'Mean Squared Error of testing dataset is {rf_mse}')
```

## 13.3 Evaluation after Hyperparameter Tuning

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_pred = rf_best_pipeline.predict(X_test)
print(f'R2 score for random forest after hyperparameter tuning {r2_score(y_test, rf_best_pred)}')
print(f'Mean Absolute Error for random forest after hyperparameter tuning {mean_absolute_error(y_test, rf_best_pred)}')
```

## 13.4 RMSE

```python
# For Linear Regression Model
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred))

# For Random Forest Model
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))

# For Tuned Random Forest Model
rf_tuned_pred = random_search.best_estimator_.predict(X_test)
rf_tuned_rmse = np.sqrt(mean_squared_error(y_test, rf_tuned_pred))
rf_tuned_training_pred = random_search.best_estimator_.predict(X_train)
rf_tuned_training = r2_score(y_train, rf_tuned_training_pred)
rf_tuned_mae = mean_absolute_error(y_test, rf_tuned_pred)
rf_tuned_r2 = r2_score(y_test, rf_tuned_pred)

# Create a DataFrame to consolidate results
results = pd.DataFrame({
    'Models': ['Linear Regression', 'Random Forest (Default)', 'Random Forest (Tuned)'],
    'R2': [lr_r2, rf_r2_test, rf_tuned_r2],
    'Training R2': [lr_r2_train, rf_r2_train, rf_tuned_training],
    'MAE': [lr_mae, rf_mae, rf_tuned_mae],
    'RMSE': [lr_rmse, rf_rmse, rf_tuned_rmse]
})

print(results)
```
